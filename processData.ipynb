```python
import pandas as pd

def contains_all_words(text, words):
    """Check if all words are present in text (any order). Case insensitive."""
    text_lower = str(text).lower()
    return all(word in text_lower for word in words)

for counter, full_name in enumerate(npk_excluded_raw_source.iloc[:, 4].tolist()):
    words = full_name.split()
    
    # Filter rows where 'name' column contains all words (in any order)
    filtered = df[df['name'].apply(lambda x: contains_all_words(x, words))]
    
    if filtered.empty:
        continue  # skip if no matches
    
    row = {
        "code": npk_excluded_raw_source.iloc[counter, 2],
        "name": full_name,
        "avg": filtered['price_per_kg'].mean(),
        "min": filtered['price_per_kg'].min(),
        "max": filtered['price_per_kg'].max(),
        "std": filtered['price_per_kg'].std(),
        "listOfNames": filtered['name'].tolist(),
        "listOfPrices": filtered['price_per_kg'].tolist(),
    }
    rows.append(row)

result_df = pd.DataFrame(rows)
print(result_df)
```

***Remove Irrelevent Price Values***

```python
def remove_outliers_iqr(prices, names, k=1.5):
    """
    Remove outlier prices and corresponding names using the IQR method.
    prices: list of floats
    names: list of strings (same length as prices)
    Returns filtered prices and names.
    """
    if not prices or len(prices) < 4:
        # Not enough data to filter, return original
        return prices, names

    arr = np.array(prices)
    q1 = np.percentile(arr, 25)
    q3 = np.percentile(arr, 75)
    iqr = q3 - q1
    lower_bound = q1 - k * iqr
    upper_bound = q3 + k * iqr

    filtered_prices = []
    filtered_names = []
    for price, name in zip(prices, names):
        if lower_bound <= price <= upper_bound:
            filtered_prices.append(price)
            filtered_names.append(name)

    return filtered_prices, filtered_names

cleaned_data = []

for idx, row in result_df.iterrows():
    prices = row['listOfPrices']
    names = row['listOfNames']

    filtered_prices, filtered_names = remove_outliers_iqr(prices, names)

    if filtered_prices:
        avg = np.mean(filtered_prices)
        minimum = np.min(filtered_prices)
        maximum = np.max(filtered_prices)
        std = np.std(filtered_prices, ddof=1) if len(filtered_prices) > 1 else 0.0
    else:
        avg = minimum = maximum = std = np.nan
        filtered_names = []

    cleaned_data.append({
        "code": row['code'],
        "name": row['name'],
        "avg": avg,
        "min": minimum,
        "max": maximum,
        "std": std,
        "listOfNames": filtered_names,
        "listOfPrices": filtered_prices,
    })

cleaned_df = pd.DataFrame(cleaned_data)
print(cleaned_df)
```

```python
def mad_based_filter(prices, names, thresh=3.5):
    """
    Remove outliers using Median Absolute Deviation (MAD).
    prices: list of floats
    names: list of strings (same length)
    thresh: threshold for modified z-score to consider outlier (usually 3-3.5)
    Returns filtered prices and names.
    """
    if not prices or len(prices) < 3:
        # Not enough data to filter robustly, return as is
        return prices, names

    arr = np.array(prices)
    median = np.median(arr)
    mad = np.median(np.abs(arr - median))

    if mad == 0:
        # MAD is zero means no variation, return original
        return prices, names

    modified_z_scores = 0.6745 * (arr - median) / mad

    filtered_prices = []
    filtered_names = []
    for price, name, z in zip(prices, names, modified_z_scores):
        if np.abs(z) <= thresh:
            filtered_prices.append(price)
            filtered_names.append(name)

    return filtered_prices, filtered_names

cleaned_data = []

for idx, row in result_df.iterrows():
    prices = row['listOfPrices']
    names = row['listOfNames']

    filtered_prices, filtered_names = mad_based_filter(prices, names)

    if filtered_prices:
        avg = np.mean(filtered_prices)
        minimum = np.min(filtered_prices)
        maximum = np.max(filtered_prices)
        std = np.std(filtered_prices, ddof=1) if len(filtered_prices) > 1 else 0.0
    else:
        avg = minimum = maximum = std = np.nan
        filtered_names = []

    cleaned_data.append({
        "code": row['code'],
        "name": row['name'],
        "avg": avg,
        "min": minimum,
        "max": maximum,
        "std": std,
        "listOfNames": filtered_names,
        "listOfPrices": filtered_prices,
    })

cleaned_df = pd.DataFrame(cleaned_data)
print(cleaned_df)
```
